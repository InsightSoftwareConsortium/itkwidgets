```python
import SimpleITK as sitk
import numpy as np
from itkwidgets import view
```

## Create a Simple 3D Image

Let's create a synthetic 3D image with SimpleITK:

```python
# Create a 3D image using SimpleITK's GaussianSource
size = [64, 64, 64]
image = sitk.GaussianSource(
    sitk.sitkFloat32,
    size,
    sigma=[10.0, 10.0, 10.0],
    mean=[32.0, 32.0, 32.0],
    scale=255.0,
    normalized=False
)

# Set spacing and origin
image.SetSpacing([1.0, 1.0, 2.0])  # Anisotropic spacing
image.SetOrigin([0.0, 0.0, 0.0])

print(f"Image size: {image.GetSize()}")
print(f"Image spacing: {image.GetSpacing()}")
print(f"Image origin: {image.GetOrigin()}")
```

## Visualize the SimpleITK Image

Now we can directly pass the SimpleITK image to itkwidgets:

```python
view(image)
```
